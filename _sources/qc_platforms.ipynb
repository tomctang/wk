## 回测Backtest 模拟(实盘) 交易live trading

2种回测系统 Vector(speed) .VS. Event(real), 各1套
- 向量化回测框架
    - 优点：1简：策略逻辑可抽象成 pandas/numpy 数据处理; 2快：一次性处理整段时间序列，适合策略参数搜索; 3适合研究阶段的 Alpha 策略验证
    - 缺点：1无法模拟实际撮合细节（委托簿、滑点、延迟）; 2不支持异步事件（如宏观信号、异步市场数据）= 未来函数风险大?; 3难对接真实交易系统
    - 适用策略: 趋势跟踪、因子选股
- 事件驱动回测/实盘框架
    - 优点：1贴近实盘：逐tick/逐bar处理，考虑委托/撮合/滑点; 2支持复杂交易逻辑（如组合管理、风险控制、订单管理系统）; 3和实盘无缝对接
    - 缺点：1慢; 2开发难度大：需要维护事件队列、撮合机制、账户模型
    - 适用策略: 高频交易、做市、套利
    - 支持 Bar → Signal → Order → Execution

```
# backtesting_compare: Vector.VS. Event
import numpy as np
import pandas as pd

"""
Minimal demo to compare a vectorized backtester and an event‑driven backtester
running the SAME simple‑moving‑average (SMA) crossover strategy.

Run this file (`python backtesting_compare.py`) to print performance metrics
for both engines. The data is synthetic so the script is self‑contained.
"""

###############################################################################
# Utility helpers
###############################################################################

def generate_data(seed: int = 42, periods: int = 250) -> pd.DataFrame:
    """Create a fake price series so the demo is entirely offline‑capable."""
    rng = np.random.default_rng(seed)
    dates = pd.date_range("2024-01-01", periods=periods, freq="D")
    prices = 100 + rng.standard_normal(periods).cumsum()
    return pd.DataFrame({"Close": prices}, index=dates)

def perf_metrics(equity_curve: pd.Series) -> dict:
    """Return Total Return, Sharpe and Max Drawdown for an equity curve."""
    rets = equity_curve.pct_change().dropna()
    total_return = equity_curve.iloc[-1] / equity_curve.iloc[0] - 1
    sharpe = rets.mean() / rets.std() * np.sqrt(252) if rets.std() != 0 else np.nan
    running_max = equity_curve.cummax()
    drawdown = (equity_curve - running_max) / running_max
    max_dd = drawdown.min()
    return {
        "total_return": round(total_return, 4),
        "sharpe": round(sharpe, 2),
        "max_drawdown": round(max_dd, 4),
    }

###############################################################################
# 1) Vectorized backtester (pandas‑style)
###############################################################################

def vectorized_backtest(
    df: pd.DataFrame, fast: int = 10, slow: int = 30, initial_capital: float = 1_000_000
):
    data = df.copy()
    data["fast"] = data["Close"].rolling(fast).mean()
    data["slow"] = data["Close"].rolling(slow).mean()

    # Trading signal: long when fast MA > slow MA; flat otherwise
    data["position"] = (data["fast"] > data["slow"]).astype(int).shift().fillna(0)

    # Strategy returns (no transaction cost, perfect fills)
    data["strategy_ret"] = data["position"] * data["Close"].pct_change().fillna(0)
    data["equity"] = (1 + data["strategy_ret"]).cumprod() * initial_capital

    return perf_metrics(data["equity"]), data

###############################################################################
# 2) Event‑driven backtester (single‑asset, bar‑by‑bar loop)
###############################################################################

def event_backtest(
    df: pd.DataFrame, fast: int = 10, slow: int = 30, initial_capital: float = 1_000_000
):
    price = df["Close"].values
    fast_ma = pd.Series(price).rolling(fast).mean().values
    slow_ma = pd.Series(price).rolling(slow).mean().values

    equity_curve = []
    cash, position = initial_capital, 0.0  # dollars, shares

    for t in range(len(price)):
        # generate signal when both MAs are ready
        if t >= slow:
            # Cross‑up
            if (
                position == 0
                and fast_ma[t - 1] <= slow_ma[t - 1]
                and fast_ma[t] > slow_ma[t]
            ):
                # BUY at open price of bar t (here use close for simplicity)
                position = cash / price[t]
                cash = 0
            # Cross‑down
            elif (
                position > 0
                and fast_ma[t - 1] >= slow_ma[t - 1]
                and fast_ma[t] < slow_ma[t]
            ):
                # SELL everything
                cash = position * price[t]
                position = 0

        # Equity value mark‑to‑market at bar close
        equity = cash + position * price[t]
        equity_curve.append(equity)

    equity_ser = pd.Series(equity_curve, index=df.index)
    return perf_metrics(equity_ser), equity_ser

###############################################################################
# Demo runner: compare both engines on the same data & strategy
###############################################################################

def main():
    df = generate_data()

    vec_metrics, vec_df = vectorized_backtest(df)
    evt_metrics, evt_curve = event_backtest(df)

    print("=== Vectorized Backtest ===")
    for k, v in vec_metrics.items():
        print(f"{k}: {v}")

    print("\n=== Event‑Driven Backtest ===")
    for k, v in evt_metrics.items():
        print(f"{k}: {v}")

    print("\nDelta (Vectorized – Event):")
    for k in vec_metrics:
        print(f"{k}: {vec_metrics[k] - evt_metrics[k]:.4f}")

if __name__ == "__main__":
    main()
```

**自动交易系统(Automated Trading System, ATS)** 一套可以自动执行交易策略的计算机程序
   - 市场数据模块（Market Data） → 获取行情数据（价格、成交量、订单簿等）
   - 交易策略模块（Trading Algorithm） → 计算买卖信号
   - 订单执行模块（Order Execution） → 发送交易订单（下单、撤单、修改订单）
   - 风险管理模块（Risk Management） → 控制亏损、监控市场（止损、止盈、仓位管理）
   - 日志 & 监控模块（Logging & Monitoring） → 记录交易 & 复盘, 改进策略

1. 选择交易市场 & 券商: 券商必须提供 API --> 获取市场数据, 分析行情
   - 美股：Interactive Brokers（IB）、TD Ameritrade, TradeStation
   - 期货 & 外汇：Oanda、NinjaTrader、TradeStation
   - 加密货币：Binance、FTX、Kraken
2. 运行交易策略, 生成买卖信号, 交易策略的核心组成: 
   - 信号生成（Signal Generation）：例如均线交叉（10 日均线上穿 50 日均线买入）。
   - 仓位管理（Position Sizing）：计算买入多少股（如最多 10% 账户资金）。
   - 风险控制（Risk Management）：设置止损（-5% 卖出），防止亏损过大。

4. 订单管理（Order Management）, 当策略决定买卖后，系统需要发送交易订单
   - 市价单（Market Order）|	立即成交	|适合高流动性资产
   - 限价单（Limit Order）	|只在特定价格成交	|适合低流动性市场
   - 止损单（Stop Loss）|	亏损超过一定数值后自动卖出	| 适合风险控制

1. 实时监控 & 风险管理: 防止异常情况（滑点、订单失败）。监控哪些风险？
   - 网络问题|	交易失败	|解决：备用服务器
   - 极端市场情况|	突破止损|解决：预设“熔断”机制
   - 账户资金不足|	无法下单|	解决：资金管理系统

2. 日志 & 交易记录: 每笔交易必须记录下来，方便复盘和优化策略。
   - 交易日志包括: 1 交易时间; 2 买卖信号; 3 成交价格; 4 盈利 & 亏损
