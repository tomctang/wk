**自动交易系统(Automated Trading System, ATS)**

一套可以自动执行交易策略的计算机程序

- 市场数据模块（Market Data） → 获取行情数据（价格、成交量、订单簿等）
- 交易策略模块（Trading Algorithm） → 计算买卖信号
- 订单执行模块（Order Execution） → 发送交易订单（下单、撤单、修改订单）
- 风险管理模块（Risk Management） → 控制亏损、监控市场（止损、止盈、仓位管理）
- 日志 & 监控模块（Logging & Monitoring） → 记录交易 & 复盘, 改进策略


1. 选择交易市场 & 券商: 差异在于交易规则、数据获取方式、执行速度等

市场	| 适用交易	| 主要特点
|--|--|--|
股票（美股、A股）|	价差交易、趋势跟踪	|交易时间固定，波动较大
期货（商品、股指）|	套利、趋势交易	|有杠杆，保证金交易
外汇（Forex）	|高频、趋势交易	|24 小时交易，低成本
加密货币（Crypto）|	高频、套利	|24 小时交易，波动大

券商: 券商必须提供 API

- 美股：Interactive Brokers（IB）、TD Ameritrade, TradeStation
- 期货 & 外汇：Oanda、NinjaTrader、TradeStation
- 加密货币：Binance、FTX、Kraken

2. 获取市场数据, 分析行情

数据类型|	作用	|来源/获取方式
--|--|--
历史数据|	回测策略，分析长期趋势|	Yahoo Finance, Quandl, Binance. 数据库查询（MySQL, PostgreSQL）
实时行情（L1 数据）|	计算技术指标，生成买卖信号|	IB API, Binance API
订单簿数据（L2 数据）|	高频交易策略使用 |	深度流动性池（Dark Pool）
财务数据 & 新闻数据|	基本面选股	|Quandl, Alpha Vantage. 爬取网页数据（Scrapy, Selenium）

3. 运行交易策略, 生成买卖信号, 交易策略的核心组成: 
   - 信号生成（Signal Generation）：例如均线交叉（10 日均线上穿 50 日均线买入）。
   - 仓位管理（Position Sizing）：计算买入多少股（如最多 10% 账户资金）。
   - 风险控制（Risk Management）：设置止损（-5% 卖出），防止亏损过大。

4. 订单管理（Order Management）, 当策略决定买卖后，系统需要发送交易订单

订单类型|	说明	|适用情况
--|--|--
市价单（Market Order）|	立即成交	|适合高流动性资产
限价单（Limit Order）	|只在特定价格成交	|适合低流动性市场
止损单（Stop Loss）|	亏损超过一定数值后自动卖出	| 风险控制

5. 实时监控 & 风险管理: 防止异常情况（滑点、订单失败）。监控哪些风险？

监控项	| 可能风险	|解决方案
--|--|--
网络问题|	交易失败	|备用服务器
极端市场情况|	突破止损|	预设“熔断”机制
账户资金不足|	无法下单|	资金管理系统

交易日志包括: 1 交易时间; 2 买卖信号; 3 成交价格; 4 盈利 & 亏损
