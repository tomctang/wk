# Terms

1. 趋势策略（股票突破新高买入）
    - 均衡打破, 价格涨了就继续涨，跌了就继续跌
        - eg.股票破新高--某股票过去6个月一直涨，你决定买入，因为它可能继续上涨。直到趋势反转（如股价跌破50日均线）卖出
    - 适合: 小盘股、成长股、加密货币（这些资产容易出现大趋势）
    - 风险: 市场趋势突然反转eg.政策变动、财报暴雷，你的交易可能会被“套牢”。市场上竞争者增加, 回归均衡, 趋势结束
2. 均值回归策略（套利策略）
    - 短期内, 涨太多会跌回均衡值；跌太多会涨回去。
        - eg.股票的布林带（Bollinger Bands）, 某股票过去 100 天的平均价格是 $100，但今天跌到了 $90，你认为它太便宜了，未来可能涨回去，于是买入。当它涨回100以上卖出获利
    - 适合: 震荡市（价格在一个区间内上下波动）。 波动大但长期稳定的市场（eg.大盘股，ETF, crypto）
    - 风险: 如果市场趋势发生变化，比如股票的“合理价格”下降了，你可能会误以为它会反弹，结果继续下跌
3. 策略/因子表达式: 即**Alpha**, 不断拷证alpha的稀缺性/背后的逻辑/规模和持久性
    - Alpha is usually expressed as a number that refers to a percentage. For example, a mutual fund with an alpha of 2.0 would mean that fund has outperformed the market average by 2%
    - Alpha的典型来源: 每个回报来源都来自一个风险源，因此任何投资组合都可以被看作是风险的组合
        - 价格/成交量
        - 基本面
        - 宏观数据
        - 文本：如期刊、新闻、会议记录、社交媒体信息
4. 夏普比率 (Sharpe Ratio)
   - 风险度量: 风险通常用收益率的波动性来表示，波动性越大，风险也就越高。标准差是衡量波动性的一种指标。eg.一个投资组合的年化收益率波动为10%，这10%就代表了你在一年内可能遇到的正负波动幅度。
   - 超额收益: 指的是投资收益减去无风险收益（比如国债收益率）的部分。它反映了你因承担风险而获得的额外收益。
   - 单位风险能赚多少钱: 超额收益/风险（标准差），就得到了每承担一单位风险所获得的超额收益。
   - 收益的波动程度（标准差）Sharpe Ratio = 策略的年收益 - 无风险收益
        - 夏普比率 < 1 ：风险太大，波动太猛，不能单独用。
        - 夏普比率 > 2 ：策略比较稳健，基本能每个月赚钱(每承担1单位的风险（标准差），你可以获得2单位的超额收益)
        - 夏普比率 > 3 ：策略非常稳定，可能是日内高频交易策略，每天都能赚钱。
    - 组合A的表现更好, SR更高:
       - 投资组合A：超额收益为6%，标准差为5%，则夏普比率为 6%/5%=1.2。eg.每个月赚 3%，但每个月都稳定盈利，从不大赚大亏
       - 投资组合B：超额收益为8%，标准差为8%，则夏普比率为 8%/8%=1.0。eg.每个月赚 5%，但有时候赚 10%，有时候亏 5%，波动很大
5. 最大回撤（Maximum Drawdown, MDD）可以用最简单的话来理解：它表示在一段时间内，你的投资或账户价值从最高点到最低点的最大下降幅度
   - 最大亏损幅度：比如你账户从最高点 10 万跌到 7 万，就是 30% 回撤。
   - 最长亏损持续时间：比如这次亏损花了 6 个月才恢复，你能熬得住吗？
6. Kelly formula - 资金管理数学方法，以获得最大化复利增长。计算: 1) 最优杠杆; 2) 资金分配
    - 如果一个策略的平均收益率（期望收益）是 𝑚，标准差（波动性）是 s，那么最优杠杆（该策略的资金比例） 𝑓 = 𝑚/𝑠^2
    - SP500 ETF（SPY）：计算得用 2.5 倍杠杆来投资 SPY，以获得最大复利增长
        - 平均年收益：11.23%
        - 年化波动率（标准差）：16.91%
        - 风险利率：4%
        - Kelly： 𝑓 =（11.23%−4%）/16.91*16.91=2.528
7. 股票的布林带（Bollinger Bands）
